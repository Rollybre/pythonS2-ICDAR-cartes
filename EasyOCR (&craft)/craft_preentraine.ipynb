```python
from craft_text_detector import Craft
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import os
os.chdir(r"C:\Users\Power Media\Documents\@ Fac\2024-2025 M2 HN Chartes\Python\evaluation")

# Charger le modèle
craft = Craft(output_dir='craft_output', crop_type="poly", cuda=False)
```

```python
# Lire l'image
image_path = 'ign/val/000001.jpg'
prediction_result = craft.detect_text(image_path)

# Charger l'image originale (PIL)
img = Image.open(image_path)
img_np = np.array(img)

# Afficher l'image avec les polygones détectés
fig, ax = plt.subplots(figsize=(12, 12))
ax.imshow(img_np)

for poly in prediction_result['polys']:
    poly = np.array(poly)
    # Créer un patch polygone rouge sans remplissage
    patch = plt.Polygon(poly, closed=True, edgecolor='red', fill=False, linewidth=2)
    ax.add_patch(patch)

plt.axis('off')
plt.show()


# Nettoyer si besoin
craft.unload_craftnet_model()
craft.unload_refinenet_model()
```
