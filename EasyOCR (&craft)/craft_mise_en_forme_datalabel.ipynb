```python
req_path = "CRAFT-pytorch/requirements.txt"

# Lire les lignes actuelles
with open(req_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# Remplacer torch et opencv-python par des versions récentes
new_lines = []
for line in lines:
    if line.startswith("torch=="):
        new_lines.append("torch>=2.0.0\n")
    elif line.startswith("torchvision=="):
        new_lines.append("torchvision>=0.15.0\n")  # Compatible avec torch>=2.0.0
    elif line.startswith("opencv-python=="):
        new_lines.append("opencv-python>=4.5.5\n")  # Version stable et récente
    else:
        new_lines.append(line)

# Écrire le fichier mis à jour
with open(req_path, "w", encoding="utf-8") as f:
    f.writelines(new_lines)

print("✅ requirements.txt mis à jour avec torch, torchvision et opencv-python récents")
```
