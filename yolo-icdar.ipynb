```python
## Entraînement YOLO

from ultralytics import YOLO
# Load a model

model = YOLO("yolo11n.pt")

# Train the model

model.train(data='/kaggle/input/config-kaggle/config_kaggle.yaml',
            epochs=50,
            imgsz=640,
            batch=32,  
            name='yolo_ign25synth_bbox',
            device='0,1',
            amp=True,
            workers=8,
           dropout = 0.1)
```

```python
## Training v2

model = YOLO('/kaggle/working/runs/detect/yolo_ign25synth_bbox/weights/best.pt')
model.train(data='/kaggle/input/config-kaggle/config_kaggle.yaml',
            epochs=100,
            imgsz=640,
            batch=32,  
            name='yolo_ign25synth_bbox_finetuning',
            device='0,1',
            amp=True,
            workers=8,
           dropout = 0.2)
```

```python
## Training v3

from ultralytics import YOLO 

model = YOLO('/kaggle/input/yolo-best/best.pt')
model.train(data='/kaggle/input/config-kaggle/config_kaggle.yaml',
            epochs=100,
            imgsz=640,
            batch=32,  
            name='yolo_ign25synth_bbox_finetuning',
            device='0,1',
            amp=True,
            workers=8,
           dropout = 0.2,
           degrees = 180,
            flipud = 0.5
           )
```
